Python Data types Categories
1. Basic Types

```python
a = 1 #int
b = 2.2 #float
c = 3 + 2j #complex
d = True #bool(Boolean values are True and False)
e = "hi" 'hello'#string
f = b'abcd'#bytes

print(f"a = {type(a)}\nb = {type(b)}\nc = {type(c)}\nd = {type(d)}\ne = {type(e)}\nf = {type(f)}")
```

2. Container Types

```python
a = [10,20,'hi',"hello",2.2,2.4,] #list
b = (10,2.6,'hi') #tuple
c = {20,22.6,'hello'} #set
d = {'key':'value','name-1':'tanvi'} #dictionary

print(f"{a}\n{b}\n{c}\n{d}")
```

3. User-defined Types

```python
class student:
    def set_data(self,n,a):
        self.name = n
        self.age = a
        
    def display_data(self):
        print(f"Name : {self.name}\nAge : {self.age}")
        
s1 = student()
s1.set_data('tanvi',17)
s1.display_data()
```

Arithmetic Operators

```python
a = 10; b = 3
print("add : ",a+b)
print("sub : ",a-b)
print("multi : ",a*b)
print("div : ",a/b)
print("modulo : ",a%b)
print("power of or raise to : ",a**b)
print("floor div : ",a//b)
```
