String index

```python
msg = 'She said, "i am tanvi"'
print("1 : ",msg[5])
print("2 : ",msg[:])
print("3 : ",msg[0:5])
print("4 : ",msg[:5])
print("5 : ",msg[5:])
print("6 : ",msg[1:8])
print("8 : ",msg[1:-8])
print("9 : ",msg[-5:])
print("10: ",msg[:-5])
print("11: ",msg[::1])
print("12: ",msg[::2])
print("13: ",msg[::-1])
print("14: ",msg[::])
```
