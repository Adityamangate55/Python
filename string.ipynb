String negetive and positive index

```python
msg = 'She said, "i am tanvi"'
print("1 : ",msg[5])
print("2 : ",msg[:])
print("3 : ",msg[0:5])
print("4 : ",msg[:5])
print("5 : ",msg[5:])
print("6 : ",msg[1:8])
print("8 : ",msg[1:-8])
print("9 : ",msg[-5:])
print("10: ",msg[:-5])
print("11: ",msg[::1])
print("12: ",msg[::2])
print("13: ",msg[::-1])#Reverse String
print("14: ",msg[::])
```

String Properties

```python
msg1 = "Hello"
msg2 = "Tanvi"

print(msg1 + msg2)
print(msg1 + " " + msg2)
print(msg1 + " Tanvi_")
print(msg1 * 2)
print("el" in msg1)#use of 'in' to check "e" present in string
print("a" in msg1)
print("eo" in msg1)
```

String Methods Part-1

```python
msg1 = "        hello Tanvi mangate    "

print("capitalize() : ",msg1.capitalize())#makes only first letter of first word capital
print("upper() : ",msg1.upper())#all words in upper case
print("lower() : ",msg1.lower())#all words in lower case
print("title() : ",msg1.title())#makes all first letter of word capital
print("lstrip() : ",msg1.lstrip())#To remove left white-spaces
print("rstrip() : ",msg1.rstrip())#To remove right white-spaces
print("strip() : ",msg1.strip())#To remove white-spaces
print("swapcase() : ",msg1.swapcase())#It invert case 
print("replace() : ",msg1.replace("h","m"))
print("replace() : ",msg1.replace("hello","hi"))
print("replace() : ",msg1.replace(" ","_+"))
print("replace() : ",msg1.replace(" ",""))
print("replace() : ",msg1.replace(" ",""))
print("count() : ",msg1.count("l"))#Count of substring
print("count() : ",msg1.count("l",10,11))
#index(substring,[start_index],[end_index])
print("index() : ",msg1.index("hello"))
print("index() : ",msg1.index("a"))
print("index() : ",msg1.index("a",16))
print("index() : ",msg1.index("a",10,16))
#find(substring,[start_index],[end_index])
#Return -1 if substring not found
print("find() : ",msg1.find("a"))
print("find() : ",msg1.find("H"))
print("split() : ",msg1.split())

msg2 = ['Hi','Tanvi']
msg3 = "hi tanvi"
print("join() : ","_".join(msg2))
print("endswith() : ",msg3.endswith("hi"))
print("endswith() : ",msg3.endswith("tanvi"))
print("startswith() : ",msg3.startswith(""))
print("startswith() : ",msg3.startswith("tanvi"))
```
